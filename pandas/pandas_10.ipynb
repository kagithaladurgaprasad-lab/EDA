# apply()

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
data={
    'Name':['Arjun','Bob','Cherry','Dhanush'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Joining_year':[2019,2020,2018,2021]
}
df=pd.DataFrame(data)
df
```

```python
df['Upadated_salary']=df['Salary'].apply(lambda x:x+5000)
df
```

```python
def salary_category(Salary):
    if Salary>=60000:
        return 'High'
    else:
        return 'Low'

df['Salary_category']=df['Salary'].apply( salary_category)
df
```

```python
import datetime

current_year=datetime.datetime.now().year
df['Experiance']=df.apply(lambda row :current_year-row['Joining_year'],axis=1)
df
```

```python
df['Name_upper']=df["Name"].apply(str.upper)
df
```

```python
df['Dept_len']=df['Department'].str.len()
df
```

```python
def add_sal(Salary):
    Salary+=5000
    return Salary

df['Salary_add']=df['Salary'].apply( add_sal)
df
```

```python
data={
    'Name':['Arjun','Bob','Cherry','Dhanush'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Joining_year':[2019,2020,2018,2021]
}
df=pd.DataFrame(data)
df
```

```python
df['Updated_salary']=df['Salary'].map(lambda x:x+5000)
df
```

```python
#map department names to codes
dept_map={'HR':'D1','IT':'D2','Finance':'D3'}
df['Dept_code']=df['Department'].map(dept_map)
df
```

```python
bonus=pd.Series([500,600,700,800],index=[0,1,2,3])
df['Bonus']=df.index.map(bonus)
df
```

```python
data={
    'Salary':[50000,60000,55000,70000],
    'Numbers':[2019,2020,2018,2021]
}
df=pd.DataFrame(data)
df
```

```python
data={
    'Name':['Arjun','Bob','Cherry','Dhanush'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Bouns':[5000,6000,5500,6000]
}
df=pd.DataFrame(data)
df
```

```python
def add_numeric(x):
    if isinstance(x,(int,float)):
        return x+1000
    return x
df_add=df.applymap(add_numeric)
df_add
```

```python
def col_upp(x):
    if isinstance(x,str):
        return x.upper()
    return x
df_upp=df.applymap(col_upp)
df_upp
```

```python
df_upper=df.applymap(lambda x:x.upper() if isinstance(x,str) else x)
df_upper
```

```python
def numeric(x):
    if isinstance(x,(int,float)):
        if x>60000:
            return 'High'
    return x
df3=df.applymap(numeric)
df3
```

# Rolling() and Expanding()

```python
sales=pd.Series([10,20,30,40,50])
sales
```

```python
print("Rolling:\n",sales.rolling(3).mean())
print('----------------------------')
print("Expanding:\n",sales.expanding().mean())
```
