# Line Plot

1.it is a graph that connects data points with a line
2.it is used to show trends or changes over time

```python
x=['jan','feb','mar','apr','may']
y=[400,500,600,450,900]
plt.plot(x,y,marker='o',linestyle='-',color='g',label='Line 1')
plt.title('Line Plot Example')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
```

# Area Plot()

it is similar to line plot,but the area under the line is filed with color 

```python
x=['jan','feb','mar','apr','may']
y1=[400,500,600,450,900]
y2=[1,1,1,1,1]
plt.fill_between(x,y1,y2, color='skyblue',alpha=0.8,label='Area plot')
plt.title('Line Plot Example')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

# Bar Plot

It is graph that uses rectunglur bars to show values 
the length or height of each bar represents the value of category
compare values accross different categories

```python
x=['Laptop','Mobile','Earphone','Tablets','Headphones']
y=[500,300,600,250,100]

plt.bar(x,y,color='b',alpha=0.6)
plt.title('Bar Plot Example')
plt.xlabel('Electronic Gadgets')
plt.ylabel('Sales')
plt.grid(axis='y')
plt.show()
```

# scatter plot

Shows the relationship b/w two variables

```python
x=[1,2,3,4,5]
y=[10,14,18,16,12]
plt.scatter(x,y,marker='o',color='r',label='Data Points')
plt.title('scatter Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

# Pie Chart

It is a circular chat that shows proportions/ percentages of categories
each slice of the circle represents a category's contribution to the whole

```python
x=['Laptop','Mobile','Earphone','Tablets','Headphones']
y=[20,15,25,30,10] #percentages

plt.pie(y,labels=x, autopct='%1.1f%%', startangle=90,colors=['gold','lightgreen','skyblue'])
plt.title("pie char example")
plt.axis('equal')
plt.show()

```

```python


x=['Laptop','Mobile','Earphone','Tablets','Headphones']
y=[20,15,25,30,10] #percentages

explode=(0,0.2,0,0,0) #pull slice from pie chart
#0--->slice in same place 
#0> --->slice moves outward


plt.pie(y,labels=x, autopct='%1.1f%%', startangle=90,colors=['gold','lightgreen','skyblue'],explode=explode,shadow=True)
plt.title("pie char example",fontsize=14,fontweight='bold')
plt.axis('equal')
plt.show()

```

# Histogram

1.deivide data into ranges (called bins)
2.counts how many values fall in each bin
3.displays bars representing those counts

```python
data=[10,15,15,20,20,20,25,25,30]
plt.hist(data,bins=5,color='green',alpha=0.7)
plt.title('Histogram Example')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```
