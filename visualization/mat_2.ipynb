```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv("Data.csv")
df
```

```python
plt.figure(figsize=(5,4)) #width and height
plt.plot(df['Name'],df['Age'],marker='o',linestyle='-',color='g',label='Line 1')
plt.title('Line Plot Example')
plt.xlabel('Name')
plt.ylabel('Age')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python

plt.figure(figsize=(5,3))
plt.fill_between(df['Name'],df['Salary'] ,color='skyblue',alpha=0.8)
plt.plot(df['Name'],df['Salary'],marker='o',color='g')
plt.title('Line Plot Example')
plt.xlabel('Employee')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python

plt.bar(df['Name'],df['Salary'],color='g',alpha=0.6)
plt.title('Bar Plot Example')
plt.xlabel('Name')
plt.ylabel('Salary')
plt.grid(axis='y')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.scatter(df['Salary'],df['Rating'],marker='o',color='r',label='Data Points')
plt.title('Salary vs Rating')
plt.xlabel('Salary')
plt.ylabel('Rating')

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.hist(df['Age'],bins=3,color='purple',alpha=0.7)
plt.title('Age distribution')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
explode=[0.1]+[0]*(len(df['Salary'])-1)  # 0.1 for first slice ,rest 0
plt.pie(df['Salary'],labels=df['Name'], autopct='%1.1f%%', startangle=90,explode=explode,shadow=True,colors=plt.cm.Set3.colors)
plt.title("Salary Ditribution",fontsize=14,fontweight='bold')
plt.axis('equal')
plt.show()
#plt.cm--->color map module
#Set3-->categorical colormap--->contains bright and visually distnict colors 
#.colors--->list of colors
```

```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
tips.shape
```

```python
#univariate plot -->we need to take only one axis as input(x or y)
```

```python
#histogram  --> univariate 
sns.histplot(data=tips,x='total_bill',bins=20,kde=True)  #kernal density estimation
plt.title("total bill")
plt.show()
```

```python
sns.histplot(data=tips,x='total_bill',bins=20,color='skyblue',edgecolor='red',kde=True)
plt.xlim(0,60) # x-axis limit
plt.ylim(0,50)
plt.title("Histogram",fontsize=16)
plt.xlabel("total bill",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

# kde plot

insted of drawing bars like Histogram.
KDE plot draws a smooth curve that represents how data points are distributed

```python
sns.kdeplot(data=tips ,x='total_bill',fill=True)
plt.xlim(0,60) # x-axis limit
plt.ylim(0,0.06)
plt.title("KDE plot of total bill")
plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

## Box Plot

```python
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("Box plot of total bill")
plt.show()
```
